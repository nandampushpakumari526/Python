```python
import matplotlib.pyplot as plt
```

```python
x=[2,4,6,7,8]
y=[1,2,3,4,5]
y1=[5,6,7,8,9]
y2=[6,7,8,9,10]
y3=[2,3,4,5,6]
plt.plot(x,y)
plt.suptitle("Data")
plt.title("Graph-1")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
plt.subplot(2,2,1)
plt.suptitle("Data")
plt.title("Graph-2")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.plot(x,y)
plt.subplot(2,2,2)
plt.suptitle("Data")
plt.title("Graph-3")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.plot(x,y1)
plt.show()
```

```python
plt.subplot(2,2,3)
plt.suptitle("Data")
plt.title("Graph-4")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.plot(x,y2)
plt.subplot(2,2,4)
plt.suptitle("Data")
plt.title("Graph-5")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.plot(x,y3)
plt.show()
```
